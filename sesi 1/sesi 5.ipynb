```python
# Algoritma pseudocode 
# (kode yang seperti bahasa manusia)
# BUBBLE SORT (Sorting data secara cepat)

"""
Algoritma pseudocode
(kode yang seperti bahasa manusia)
bubble sort:
listawal : [5, 3, 8, 4, 2] 
mulai
pjglist = len(list awal)
untuk index dari 0 ke pjglist-1
    untuk jindex dari 0 ke pjglist-index-1:
           if arr[jindex] > arr[jindex+1]:
                tukar arr[jindex] dan arr[jindex + 1]
selesai



"""

def bubble_sort(data_list):
    #mulai
    pjglist = len(data_list) # ambil panjangn list 
    #lakukan looping sepanjang list 
    for index in range(0, pjglist - 1):
        #lakukan looping didalam list untuk mengecek besar kecil data 
        for jindex in range(0, pjglist - index - 1):
            # bandingkan data yang bersebelahan 
            if data_list[jindex] > data_list[jindex + 1]:
                # swap/ tukar posisinya 
                data_list[jindex], data_list[jindex + 1] = data_list[jindex + 1], data_list[jindex]

                """
                data_list[jindex] = data_list[jindex + 1]
                data_list[jindex + 1] = data_list[jindex]
            
                """ 
    return data_list
    
#cara menggunakan function 
list_nilai = [80, 60, 70, 56, 89]
print("List nilai awal : ", list_nilai)
print("List Nilai Terurut :", bubble_sort(list_nilai))
```

```python
def tampilkan_produk(data):
  print("\nDaftar produk:")
  for p in produk:
    print(f"- Produk {p["nama"]}: Rp {p["harga"]}")

def bubble_sort_by_harga(data_list):
    pjglist = len(data_list) # ambil panjang list 
    for index in range(0, pjglist - 1):
        for jindex in range(0, pjglist - index - 1):
          if data_list[jindex]["harga"] > data_list[jindex + 1]["harga"]:
            data_list[jindex], data_list[jindex + 1] = data_list[jindex + 1], data_list[jindex]
    return data_list

def bubble_sort_by_nama(data_list):
    pjglist = len(data_list) # ambil panjang list 
    for index in range(0, pjglist - 1):
        for jindex in range(0, pjglist - index - 1):
          if data_list[jindex]["nama"] > data_list[jindex + 1]["nama"]:
            data_list[jindex], data_list[jindex + 1] = data_list[jindex + 1], data_list[jindex]
    return data_list

# function untuk menyatukan semua produk
def main():
   print("Selamat Datang di Toko Online Mini 👛")

   while True:
      print("\n1. Urutkan Berdasarkan Harga")
      print("2. Urutkan Berdasarkan Nama")
      print("3. Tampilkan Data")
      print("4. Keluar")

      # kita minta inputan dari user
      pilihan = input("Pilih menu (1/2/3/4): ")

      # control flow if elif else
      if pilihan == "1":
         hasil = bubble_sort_by_harga(produk)
         tampilkan_produk(hasil)
      elif pilihan == "2":
         hasil = bubble_sort_by_nama(produk)
         tampilkan_produk(hasil)
      elif pilihan == "3":
         tampilkan_produk(produk)
      elif pilihan == "4":
         print("Terima Kasih 👏")
         break
      else:
         print("Pilihan tidak valid. Coba lagi")
```
