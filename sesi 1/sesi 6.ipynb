```python
# Object dan Class
"""
Sintax kelas di python :

class NamaKelas :
    isi dari kelas
    # mengisi atribut/ data dari kelas 
    def __init__(self, atribut1, atribut2 dst)
        # self untuk mengarah ke kelas itu sendiri 
        self.atribut1 = atribut1
        dst

        # methode didalam kelas 
        def nama_method():
            isi method

            
sintax membuat object :

nama_object = NamaKelas(isiatribut1 dst)
nama_object.nama_method atau nama atribut

"""

# nama kelas

class Mobil :
    def __init__(self,merk,warna,tahun):
        self.merk = merk # atribut/data/merk
        self.warna = warna # atribut/data/ warna 
        self.tahun = tahun # atribut/data tahun


    def jalan(self):
        print(f"mobil {self.merk} sedang berjalan")

    def berhenti(self):
        print(f"mobil {self.merk} berhenti")
              
    def klakson(self):
        print(f'mobil {self.warna} klaskson')

# membuat object dari kelas 

mobil1 = Mobil("Toyota", "Hijau", 2020)
mobil2 = Mobil("Mitsubishi", "Putih", 2020)
mobil3 = Mobil("Honda", "Hitam", 2025)

# memanggil method dari kelas 
mobil1.jalan()
mobil2.jalan()
mobil3.jalan()
# challenge teman teman panggil methode berhenti dan klakson disetiap object 
mobil1.berhenti()
mobil2.berhenti()
mobil3.berhenti()

mobil1.klakson()
mobil2.klakson()
mobil3.klakson()


# panggil atribut/data

print(mobil1.warna)
print(mobil1.tahun)
```

```python
# OOP (OBJECT ORIENTED PROGRAMMING)
# Encapulation = menyembunyikan data / atribut agar tidak bisa diakses
# biasanya digunakan untuk melindungi data 

class AkunBank:
    def __init__(self, nama, saldo):
        self.nama = nama # public atribute
        self.__saldo = saldo # private saldo atribute 
        # perhatikan penggunaan __

# Setter dan getter 

    def setSaldo(self, saldo):
        self.__saldo = saldo

    def getSaldo(self):
        print(f"Saldo kamu {self.__saldo}")


# membuat object/ akun 
akun1 = AkunBank("Hafiz", 7000)
print(akun1.nama)
akun1.nama = "Anton"
print(akun1.nama)

akun1.getSaldo()

```

```python
# inheritance (pewarisan)
# -> sebuah kelas bisa mewarisi atribut dan methode dari class lain.
# mirip seperti anak mewarisi sifatnya orang tua 

# parent kelas/ super kelas 
class Hewan:
    # atributnya
    def __init__(self, nama):
        self.nama = nama 

    def suara(self):
        print("Hewan Bersuara")

    def jumlah_kaki(self):
        print("Hewan Berkaki 4")
# child kelas
class Kucing(Hewan): # kelas kucing mewarisi dari kelas hewan
    def suara(self):
        print("Meong-meong")

# child kelas
class Anjing(Hewan): # kelas anjing mewarisi dari kelas hewan
    def suara(self):
        print("guk guk")

# buat object 
mimi = Kucing("mimi")
leki = Anjing("Leki")
kucing1 = Kucing("oyen") # menambahkan nama walaupun tidak ada atribute di induknya
kucing2 = Kucing("Mujair")
mimi.suara()
leki.suara()
mimi.jumlah_kaki()
print(kucing1.nama)
```

```python
# Polimorfisme
# objek berbeda bisa memiliki methode dengan nama sama,
# tapi implementasinya berbeda.

class Person:
    def __init__(self, nama):
        self.nama = nama

    def greet(self):
        print("Hallo Teman- teman")

class English(Person):
    def greet(self): # ovveride
        print("Halo friends")

class German(Person):
    def greet(self):
        print("tchusss preunds")

# buat object 
person1 = English("Hafiz")
person2 = German("Gio")

person1.greet() # nama methode sama
person2.greet() # tapi implementasi berbeda

print(person1.nama)
psn1 = Person("tesr")
psn1.greet()
```
