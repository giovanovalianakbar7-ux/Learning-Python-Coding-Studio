```python
# BAGIAN 1 : INPUT HARGA BAHAN BAKU DIPASAR (DATABASE HARGA)
print("MASUKKAN DAFTAR BAHAN BAKU ANDA DISINI: ")
harga_terigu = 12000 # Harga per kg 
harga_tepung_beras = 9000 # Harga per bungkus 500gr
harga_tepung_ketan = 12000 
harga_gula_merah = 12000 # Harga per 500gr
harga_gula_putih = 15000
harga_mentega = 4000
harga_minyak = 17000 # per liter 
harga_wortel = 18000 # per kg
harga_bumbu = 5000
harga_gas = 22000
ubi_ungu = 10000
harga_telur = 16000 # per 250gr

# BAGIAN 2 : PEMAKAIAN BAHAN BAKU 
# Berapa gram yang terpakai dalam pemakaian bahan baku 
pakai_tepung = 2000 # dalam gram
pakai_gula_merah = 750
pakai_telur = 250
pakai_minyak = 2000
pakai_gula = 250
pakai_wortel = 2000
pakai_mentega = 1 # hitungan per bungkus
pakai_gas = 2000
pakai_tepung_beras = 500
pakai_tepung_ketan = 250
pakai_bumbu = 13
pakai_ubi_ungu = 1
jumlah_kue = 150 # banyak nya kue dalam semua resep 

# BAGIAN 3 : LOGIKA PROGRAM 
# RUMUS : (HARGA BAHAN / UKURAN BELI) * UKURAN PAKAI 

biaya_tepung = (harga_terigu / 1000) * pakai_tepung
biaya_gula_merah = (harga_gula_merah / 1000) * pakai_gula_merah 
biaya_tepung_beras = (harga_tepung_beras / 1000) * pakai_tepung_beras
biaya_tepung_ketan = (harga_tepung_ketan / 1000) * pakai_tepung_ketan
biaya_gula = (harga_gula_putih / 1000) * pakai_gula
biaya_mentega = harga_mentega * 0.5
biaya_telur = (harga_telur / 1000) * pakai_telur 
biaya_lain = 10000

# Total modal semua resep 
total = biaya_tepung + biaya_gula_merah + biaya_tepung_beras + biaya_tepung_ketan + biaya_gula 
+ biaya_mentega + biaya_telur + biaya_lain 

# HPP per 1 biji 
biji = total / jumlah_kue

# Tampilkan Hasil 
print("======LAPORAN KEUANGAN KUE======")
print(f"Total Modal 1 adonan : Rp {int (total)}")
print(f"Jumlah kue jadi : {jumlah_kue} pcs")
print(f"-" * 30)
print(f"MODAL BERSIH HPP / KUE : Rp {int(biji)}")
print("-" * 30)

# Cek Margin Keuntungan 
# Harga jual 2500 per biji 

harga_jual = 2500
keuntungan = harga_jual - biji
total_penjualan_perhari = harga_jual * jumlah_kue
omset = total_penjualan_perhari * 30
print(f"PENDAPAT PERHARI JIKA SEMUA KUE HABIS : {int(total_penjualan_perhari)}")
print(f"HARGA JUAL DI WARUNG : {harga_jual}, Untung Anda {int(keuntungan)} per kue/pcs ")
print(f"OMSET PERBULAN : Rp {int(omset)} per bulan")







```
